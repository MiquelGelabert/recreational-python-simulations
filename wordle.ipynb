```python
import random

u = ['abeto', 'abril', 'acero', 'actor', 'agudo', 'ahora', 'aireo', 'ajeno', 'alado',
 'aldea', 'alejo', 'aleta', 'aliso', 'almas', 'amado', 'amigo', 'ancho', 'antes', 'arena',
 'aroma', 'arroz', 'asado', 'asilo', 'asomo', 'astro', 'aviso', 'azote', 'babas', 'bache',
 'baila', 'bajar', 'balas', 'balon', 'barco', 'barra', 'basto', 'bebes', 'beber', 'bicho',
 'blusa', 'bolsa', 'botas', 'bravo', 'brazo', 'broma', 'burro', 'cable', 'caida', 'campo',
 'canto', 'carne', 'casas', 'cebra', 'cerco', 'cerdo', 'cerro', 'chico', 'cielo',
 'cifra', 'claro', 'clima', 'cobre', 'comer', 'copas', 'coral', 'corre', 'cruel', 'cruce',
 'cuero', 'culpa', 'curar', 'dedal', 'denso', 'dices', 'dieta', 'doble', 'dolor', 'duelo',
 'edito', 'ejote', 'elote', 'elite', 'ellos', 'enano', 'error', 'etapa', 'exodo',
 'falsa', 'feliz', 'feroz', 'ficha', 'flaco', 'flora', 'fondo',
 'forma', 'fruto', 'fuego', 'fumar', 'gafas', 'gasto', 'globo', 'gordo', 'grano', 'grupo',
 'guapo', 'hacer', 'hielo', 'hojas', 'horno', 'ideal', 'igual', 'indio',
 'jabon', 'joven', 'juego', 'jugar', 'labio', 'largo', 'lente', 'libra', 'limon', 'lucha',
 'luzco', 'mango', 'matar', 'medir', 'miedo', 'mirar', 'mismo', 'monto', 'morir', 'mujer',
 'nacer', 'negro', 'nieve', 'nubes', 'nutre', 'oasis', 'obrar', 'ocaso', 'orden',
 'padre', 'palma', 'parar', 'pasto', 'pelar', 'pesar', 'plaza', 'pobre', 'punto', 'queso',
 'quiso', 'rabia', 'raton', 'renta', 'robar', 'roble', 'ruido', 'saber', 'salto', 'silla',
 'sitio', 'sobra', 'suelo', 'tabla', 'techo', 'tenis', 'tirar', 'torre', 'traje', 'trama',
 'tumba', 'vacas', 'vapor', 'velar', 'viaje', 'vigor', 'visto', 'volar', 'yerno', 'zorro']

palabra_secreta = random.choice(u).upper()


def filtrar_palabras(verdes, amarillos, grises):
    palabras_filtradas = []
    for palabra in u:
        palabra = palabra.upper()
        cumple = True

        # Verificar letras verdes (posición exacta)
        for i, letra in verdes.items():
            if palabra[i] != letra:
                cumple = False
                break

        # Verificar letras amarillas (debe estar pero en otro lugar)
        for letra, posiciones in amarillos.items():
            if letra not in palabra or any(palabra[pos] == letra for pos in posiciones):
                cumple = False
                break

        # Verificar letras grises (no deben aparecer)
        if any(letra in palabra for letra in grises):
            cumple = False

        if cumple:
            palabras_filtradas.append(palabra)

    return palabras_filtradas[:10]  # Devolver las 10 primeras


def wordle_tramposo():
    intentos = 0
    verdes = {}  # {posicion: letra}
    amarillos = {}  # {letra: [posiciones donde NO puede estar]}
    grises = set()

    while intentos < 6:
        guess = input("Ingresa una palabra de 5 letras: ").upper()

        if len(guess) != 5:
            print("La palabra debe tener exactamente 5 letras.")
            continue

        GREEN = "\033[92m"  # Verde
        YELLOW = "\033[93m"  # Amarillo
        GRAY = "\033[90m"  # Gris
        RESET = "\033[0m"  # Reset para el color

        resultado = []
        palabra_secreta_list = list(palabra_secreta)

        # Primera pasada para verdes
        for i in range(5):
            letra = guess[i]
            if letra == palabra_secreta[i]:
                resultado.append(GREEN + letra + RESET)
                palabra_secreta_list[i] = None
                verdes[i] = letra
            else:
                resultado.append(None)

        # Segunda pasada para amarillos y grises
        for i in range(5):
            if resultado[i] is None:
                letra = guess[i]
                if letra in palabra_secreta_list:
                    resultado[i] = YELLOW + letra + RESET
                    palabra_secreta_list[palabra_secreta_list.index(letra)] = None
                    if letra not in amarillos:
                        amarillos[letra] = []
                    amarillos[letra].append(i)
                else:
                    resultado[i] = GRAY + letra + RESET
                    grises.add(letra)

        # Mostrar resultado
        print(" ".join(resultado))

        # Sugerir palabras posibles
        sugerencias = filtrar_palabras(verdes, amarillos, grises)
        print("Sugerencias:", ", ".join(sugerencias))

        if guess == palabra_secreta:
            print(GREEN + "¡Ganaste!" + RESET)
            break

        intentos += 1

    if guess != palabra_secreta:
        print(f"Perdiste, la palabra era {palabra_secreta}")

wordle_tramposo()
```
